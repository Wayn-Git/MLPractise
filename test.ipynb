```python
import numpy as np
import matplotlib.pyplot as plt 
import seaborn as sns

X = np.array([
    [1],
    [2],
    [3],
    [4],
    [5]
], dtype=float)

y = np.array([
    [3],
    [5],
    [7],
    [9],
    [11]
], dtype=float)

sns.scatterplot(x=X.flatten(), y=y.flatten())
plt.xlabel('Independent Variable')
plt.ylabel('Dependent Variable')
plt.title("Linear Relation between the data")
plt.show()
```
