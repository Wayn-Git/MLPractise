```python
import sqlite3

con = sqlite3.connect("devstash.db")

cur = con.cursor()
```

```python
cur.execute("""
    CREATE TABLE IF NOT EXISTS devstash(
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        snippet TEXT NOT NULL
    )
""")
```

```python
def add(name: str, snippet: str):
    
    cur.execute(
      "INSERT into devstash (name, snippet) VALUES (?,?)",
       (name, snippet)
    )

    con.commit()
    print("User Added Successfully")
```

```python
add("git pull", "git pull")
```

```python
def search(name: str):
    cur.execute(
        "SELECT * FROM devstash WHERE name Like ?", 
        (f"%{name}%",)
    )
    f = cur.fetchall()
    for row in f:
        print(f"ID: {row[0]}, name: {row[1]}, snippet: {row[2]}")

search("git")

```
